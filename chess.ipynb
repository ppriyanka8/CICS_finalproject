```python
def validMove_Parallel(current_square,target_square):  #parallel to the files

    a,b=current_square #unpacks coordinates of original_square into var a & b
    c,d=target_square

    if b!=d:   #to move parallely, a needs to be same as c
        print("Invalid Move")
        return False
    col=d

    if c>a:
        start=a+1
        end=c+1
        step=1

    else:
        start=a-1
        end=c-1
        step=-1

    for row in range(start,end,step):   #starts checking from a+1, because a is teh currrent position, tht will never be empty square
        if board[row][col]!=0:

            # are we at the target square? If so, if the piece belongs to opposite player, valid move. Otherwise, invalid move.
            if row==c:
                if board[a][b]>0 and board[c][d]<0:
                    print("Captured piece")
                    return True
                elif board[a][b]>0 and board[c][d]>0:
                    print("Captured piece")
                    return True
                else:  
                    return False
            else:
                return False
            
    print('can move')
    print(row, col)
    return True
```

```python
def validMove_Normal(current_square,target_square):  #takes input player=W/B

    a,b=current_square #unpacks coordinates of original_square into var a & b
    c,d=target_square

    if a!=c:
        return "Invalid move"
    row=c
    
    if d>b:
        start=b+1
        end=d+1
        step=1

    else:
        start=b-1
        end=d-1
        step=-1

    for col in range(start,end,step):

        if board[row][col]!=0:    #fault here: entering if conditon on the current sq coord
            if col==d:  #if reached the target square
                if board[a][b]>0 and board[c][d]<0:
                    print("Captured piece")
                    return True
                elif board[a][b]>0 and board[c][d]>0:
                    print("Captured piece")
                    return True
                else:  
                    return False
            else:
                return False

    print("can move")
    print(row, col)
    return True
```

```python
def validMove_Diagonal(original_square,target_square): #White: (3,2)-> (6,5)

    a,b=original_square #unpacks coordinates of original_square into var a & b
    c,d=target_square
    # current_board=board #create the updated version of board as the current_board
    if c>a and d>b:
        start=b+1
        end=d+1
        step=1

    else:
        start=b-1
        end=d-1
        step=-1

    for row in range(start,end,step):  #loops all indexes along x axis ie each column index
        for col in range(start,end,step):  #loops along y axis ie each row index
            if board[row][col]!=0:    #selects diagonal sq

                if row==c and col==d:  #if reached target sq
                    if board[a][b] * board[c][d]<0:
                        print("Captured piece")
                        return True
                    else:  
                        return False
                return False
    return True
```
